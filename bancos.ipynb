```python
import pandas as pd, yfinance as yf,numpy as np, matplotlib.pyplot as plt
bancos = {
    "Itau": "ITUB4.SA",
    "Bradesco": "BBDC4.SA",
    "Santander": "SANB11.SA",
    "Banco do Brasil": "BBAS3.SA",
}
data_inicio = "2016-01-01"
dados = yf.download(list(bancos.values()), start=data_inicio, auto_adjust=True)["Close"]
precos = dados.rename(columns = {
"ITUB4.SA": "Itau",
"BBDC4.SA": "Bradesco",
"SANB11.SA": "Santander",
"BBAS3.SA": "Banco do Brasil"
})
desempenho = precos / precos.iloc[0] * 100
resultado_final = desempenho.iloc[-1].sort_values(ascending=False)

print("\nDesempenho das ações desde a base inicial:")

for banco, valor in resultado_final.items():
    variacao = valor - 100
    print(f"{banco}: índice {valor:.1f} | variação: {variacao:+.1f}%")
fig, ax = plt.subplots(figsize=(14, 7))

for banco in desempenho.columns:
    ax.plot(
        desempenho.index,
        desempenho[banco],
        label=banco,
        linewidth=2
    )

# Eventos históricos
eventos = [
    ("2020-03-11", "Pandemia"),
    ("2021-03-17", "Alta da Selic"),
    ("2023-01-11", "Caso Americanas")
]

for data, evento in eventos:
    data = pd.Timestamp(data)

    ax.axvline(
        x=data,
        color="gray",
        linestyle="--",
        alpha=0.7,
        linewidth=1
    )

    ax.annotate(
        evento,
        xy=(data, 0.98),
        xycoords=("data", "axes fraction"),
        xytext=(5, -5),
        textcoords="offset points",
        rotation=90,
        va="top",
        ha="left",
        fontsize=9
    )

ax.set_title(
    "Desempenho histórico das ações dos principais bancos brasileiros",
    fontsize=14,
    pad=15
)
ax.set_xlabel("Data")
ax.set_ylabel("Desempenho (base 100)")
ax.legend(title="Banco")
ax.grid(True, alpha=0.25)

fig.tight_layout()
plt.show()
```
